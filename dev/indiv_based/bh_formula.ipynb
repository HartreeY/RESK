```julia
using Plots
xs = collect(0:1200)
R_PROLIF_RATE = 2
K_CAPACITY = 50

plot(xs, [x * (R_PROLIF_RATE/(1 + (x*(R_PROLIF_RATE-1))/K_CAPACITY)) for x in xs])
```

So BH actually approaches R*K. Using it without penalties and not taking into account prolif. rate means overvaluing your realistic capacity.

```julia
K_CAPACITY = 500
plot!(xs, [x * (R_PROLIF_RATE/(1 + (x*(R_PROLIF_RATE-1))/K_CAPACITY)) for x in xs])
```
