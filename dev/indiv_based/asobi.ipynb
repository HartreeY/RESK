```julia
# Disk radius
r = 30

# Generate the coordinates
coordinates = []

for x in -r:r
    for y in -r:r
        if x^2 + y^2 <= r^2
            push!(coordinates, (x, y))
        end
    end
end

# Convert the list to an array for easier usage if needed
coordinates_array = hcat([x[1] for x in coordinates], [x[2] for x in coordinates])

# Show some of the coordinates
println("First 10 coordinates:", coordinates[1:10])
```

```julia

# Plot using scatter
scatter(coordinates..., aspect_ratio=:equal, marker=:circle, markersize=2, label="Disk Points")

# Add title and labels
title!("Disk with Radius 30")
xlabel!("X-axis")
ylabel!("Y-axis")
```
