```julia
function analyse(re,name)
    re_fitn_frontav = average_front(re,"fitn";oneside=true)
    re_fitn_frontav_mean = average_ts(re_fitn_frontav, re["stats"]["n_gens"])
    re_fitn_frontav_meanN = norm_onset_mean(re_fitn_frontav_mean, re["stats"]["n_gens_burnin"]+1)
    Plots.plot!(re_fitn_frontav_meanN[(re["stats"]["n_gens_burnin"]+2):end],label="Onset mean norm. $name",xlabel="Generation")
end

analyse(A,"A")
analyse(B,"B")
#analyse(C,"C")
#analyse(D,"D")
analyse(E,"E")
```
