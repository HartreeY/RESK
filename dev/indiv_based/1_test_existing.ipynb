```julia
#using ProfileView, Profile
procs = addprocs(3)
A = rangeexp_1d_inf(0,2500,4;x_max_burnin=5,x_max_exp=500,data_to_generate="F",capacity=100,prolif_rate=2,mut_rate=0.05,migr_rate=0.05,sel_coef=0.005,weightfitn=false,condsel=true,fixed_mate=true)
```

```julia
# A is no burn-in, but since the output shouldn't change, let's test the pipeline for burn-in, to test function consistency etc.
A_fitn_frontav = average_front(A,"fitn";oneside=true)
A_fitn_frontav_mean = average_ts(A_fitn_frontav, A["stats"]["n_gens"])
A_fitn_frontav_meanN = norm_onset_mean(A_fitn_frontav_mean, A["stats"]["n_gens_burnin"]+1)
```
