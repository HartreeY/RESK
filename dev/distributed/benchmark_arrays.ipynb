```julia
using BenchmarkTools

x, y, z = 15, 15, 15

nested_array = [[[rand(Bool) for _ in 1:z] for _ in 1:y] for _ in 1:x]

flat_array = Array{Bool, 3}(undef, x, y, z)
for i in 1:x, j in 1:y, k in 1:z
    flat_array[i, j, k] = rand(Bool)
end

# Benchmark access time for nested array
@btime for i in 1:x
    for j in 1:y
        for k in 1:z
            _ = nested_array[i][j][k]  # Access elements
        end
    end
end

# Benchmark access time for flattened array
@btime for i in 1:x
    for j in 1:y
        for k in 1:z
            _ = flat_array[i, j, k]  # Access elements
        end
    end
end

# Benchmark modification time for nested array
@btime for i in 1:x
    for j in 1:y
        for k in 1:z
            nested_array[i][j][k] = rand(Bool)  # Modify elements
        end
    end
end

# Benchmark modification time for flattened array
@btime for i in 1:x
    for j in 1:y
        for k in 1:z
            flat_array[i, j, k] = rand(Bool)  # Modify elements
        end
    end
end
```
