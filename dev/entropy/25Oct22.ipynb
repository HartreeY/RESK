```julia
afs = [[af_A(A,[1],ll,gg) for ll in 1:1000] for gg in 2:131]
```

```julia
function shannon_ent(freqs)
    freqs = freqs[freqs .> 0]  # avoid log(0)
    return -sum([f*log2(f) for f in freqs])
end

shannon_ent(afs[30])
```

```julia
shannon_ent(afs[end])
```
