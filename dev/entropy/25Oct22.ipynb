```julia
function shannon_ent(_A, _a, n_loci)
    sum = 0
    for i in 1:n_loci
        if _A[i]>0
            sum-=_A[i]*log2(_A[i])
        end
        if _a[i]>0
            sum-=_a[i]*log2(_a[i])
        end
    end
    return sum/n_loci
end

shannon_ent(Afs[5,:,end],afs[5,:,end],1000)
```
