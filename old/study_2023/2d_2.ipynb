#### A 2D radial (disk) simulation

```julia
include("../resk.jl")
include("../reskplots.jl")
```

```julia
A = deserialize("data/2d_2.re")
re_heatmap(A, "fitn", A["stats"]["n_gens"]-50, A["stats"]["n_gens"])
```

```julia
re_heatmap(A,"fitn",450,500)
```

```julia
A_fitn_frontav = average_front(A,"fitn";oneside=false)
A_fitn_frontav_mean = average_ts(A_fitn_frontav, A["stats"]["n_gens"])
A_fitn_frontav_meanN = norm_onset_mean(A_fitn_frontav_mean, A["stats"]["n_gens_burnin"]+1)
Anb = deserialize("data/2d_2nb.re")
Anb_fitn_frontav = average_front(Anb,"fitn";oneside=false)
Anb_fitn_frontav_mean = average_ts(Anb_fitn_frontav, Anb["stats"]["n_gens"])
```

```julia
Plots.plot(A_fitn_frontav_mean,label="Onset mean normalisation",xlabel="Generation",ylims=(0.4,1.1),xlims=(995,1005))
```

```julia
Plots.plot(A_fitn_frontav_mean,label="Onset mean normalisation",xlabel="Generation",ylims=(0.4,1.1),xlims=((A["stats"]["n_gens_burnin"]-50),(A["stats"]["n_gens_burnin"]+50)))
vline!([A["stats"]["n_gens_burnin"]+1],label="Onset")
```

```julia
Plots.plot(A_fitn_frontav_meanN[(A["stats"]["n_gens_burnin"]+1):end],label="Onset mean normalisation",xlabel="Generation",ylims=(0.4,1.1))
Plots.plot!(Anb_fitn_frontav_mean,label="No burn-in",xlabel="Generation",ylims=(0.4,1.1))
```

```julia
Plots.plot(A_fitn_frontav_meanN[(A["stats"]["n_gens_burnin"]+2):end],label="Onset mean normalisation",xlabel="Generation",ylims=(0.4,1.1))
Plots.plot!(Anb_fitn_frontav_mean,label="No burn-in",xlabel="Generation",ylims=(0.4,1.1))
```
